```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
path="https://raw.githubusercontent.com/ZoubirCHATTI/02__Probability-Statisitics/main/01_Basis%20statistics/data/powerconsumption.csv"
df=pd.read_csv(path)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
z_consommation_zone1=(df['PowerConsumption_Zone1']-df['PowerConsumption_Zone1'].mean())/df['PowerConsumption_Zone1'].std()
```

```python
plt.figure(figsize=(15, 5))
plt.plot(z_consommation_zone1)
plt.xlabel('date index')
plt.ylabel('average consumption')
plt.title('Average electricity consumption zone one')
plt.axhline(y=3, color='r', lw=3)
plt.axhline(y=-3, color='r', lw=3)
plt.grid(True)
```

```python
z_consommation_zone2=(df['PowerConsumption_Zone2']-df['PowerConsumption_Zone2'].mean())/df['PowerConsumption_Zone2'].std()
```

```python
plt.figure(figsize=(15, 5))
plt.plot(z_consommation_zone2)
plt.xlabel('date index')
plt.ylabel('average consumption per site')
plt.title('Average electricity consumption zone two')
plt.axhline(y=3, color='r', lw=3)
plt.axhline(y=-3, color='r', lw=3)
plt.grid(True)
```
